# Meeting Minutes Generator

This notebook allows you to generate professional meeting minutes from audio recordings using various AI models.

## Features

- **Audio Transcription**: Uses OpenAI's Whisper medium.en model to transcribe audio files
- **Multiple Model Support**: Compare meeting minutes generated by different LLMs including:
  - Meta Llama 3.1/3.2 Instruct models
  - Mistral 7B Instruct
  - Qwen 2.5 Instruct models
  - DeepSeek R1 Distill models
- **Configurable Quantization**: Choose between 4-bit and 8-bit quantization with various options
- **AI-Powered Analysis**: GPT-4o-mini compares and analyzes the generated minutes
- **Performance Statistics**: View detailed performance and memory metrics for each model

## Setup Requirements

Before running this notebook, you need to:

### 1. Connect to a GPU Runtime
- Go to **Runtime** → **Change runtime type**
- Select **T4 GPU** (free tier) or higher
- Click **Save**

### 2. Set Up API Keys as Colab Secrets
You need to add two secrets to your Colab environment:

1. Click the **🔑 Secrets** icon in the left sidebar
2. Add the following secrets:
   - `HF_TOKEN`: Your Hugging Face access token ([Get one here](https://huggingface.co/settings/tokens))
   - `OPENAI_API_KEY`: Your OpenAI API key ([Get one here](https://platform.openai.com/api-keys))

### 3. Request Model Access (if needed)
Some models require approval before use. Visit these pages to request access:
- [Meta Llama models](https://huggingface.co/meta-llama)
- [Mistral models](https://huggingface.co/mistralai)
